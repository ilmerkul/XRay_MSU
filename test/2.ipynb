```python
a = 3
b = 3
c = 3
alpha = 90
beta = 90
gamma = 90

wavelength = 1.5418
```

```python
cr = Crystal(a=a, b=b, c=c, alpha=alpha, beta=beta, gamma=gamma, spacegroup_number=520)
```

```python
pp = PowderPattern(
    name="test",
    crystal=cr,
    wavelength=wavelength,
    twotheta_range=[20, 100, 0.02],
    intensity_min=0,
)
```

```python
with open("2.tsv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f, delimiter="\t")
    writer.writerow(["hkl", "d_spacing", "theta", "twotheta", "mult"])

    for ref in pp.reflections[::-1]:
        hkl = ref["hkl"]
        d = ref["d"]
        twotheta = ref["twotheta"]
        theta = twotheta / 2.0
        mult = ref["mult"]

        hkl_str = f"{hkl[0]} {hkl[1]} {hkl[2]}"

        writer.writerow([hkl_str, f"{d:.3f}", f"{theta:.3f}", f"{twotheta:.3f}", mult])
```
