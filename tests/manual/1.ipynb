```python
import csv

from src.model.crystal.crystal import Crystal
from src.model.crystal.utils import generate_hkl_by_layer
```

```python
n = 200
```

```python
a = 12.87
b = 12.87
c = 5.642
alpha = 90
beta = 90
gamma = 120
```

```python
cr = Crystal(a=a, b=b, c=c, alpha=alpha, beta=beta, gamma=gamma, spacegroup_number=1)
```

```python
with open("1.tsv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f, delimiter="\t")
    writer.writerow(["hkl", "d_spacing"])

    for hkl in generate_hkl_by_layer(n=n, include_zero=False):
        hkl_str = f"{hkl[0]:3d} {hkl[1]:3d} {hkl[2]:3d}"
        d = cr.d_spacing(hkl)
        writer.writerow([hkl_str, d])
```
